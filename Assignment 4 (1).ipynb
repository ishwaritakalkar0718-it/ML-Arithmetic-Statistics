# Experiment: Arithmetic and Statistical Calculations Using Python

## Aim
To set up a machine learning environment using Anaconda and Jupyter Notebook and perform arithmetic and statistical calculations.

## Objectives
1. Perform basic arithmetic operations.
2. Calculate mean, median and standard deviation.
3. Understand the use of Python for statistical analysis.

## Software Required
- Anaconda Distribution
- Python
- Jupyter Notebook
- NumPy and Statistics libraries

```python
# Program 1: Arithmetic Operations

a = 20
b = 10

print("First number:", a)
print("Second number:", b)
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

```python
# Program 2: Statistical Calculations

import statistics

numbers = [10, 20, 30, 40, 50]

mean_value = statistics.mean(numbers)
median_value = statistics.median(numbers)
std_deviation = statistics.pstdev(numbers)

print("Dataset:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Standard Deviation:", round(std_deviation, 2))
```

```python
# Program 3: Statistical Calculations Using NumPy

import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("Array:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", round(np.std(data), 2))
```

## Conclusion

The machine learning environment was successfully accessed using Anaconda and Jupyter Notebook. Basic arithmetic operations and statistical calculations, including mean, median and standard deviation, were performed using Python and NumPy.
